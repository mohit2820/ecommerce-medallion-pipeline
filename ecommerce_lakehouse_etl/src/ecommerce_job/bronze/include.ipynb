## CATALOG CREATION 

```python
%sql
CREATE CATALOG IF NOT EXISTS ecommerce_catalog_dev;
CREATE CATALOG IF NOT EXISTS ecommerce_catalog_uat;
CREATE CATALOG IF NOT EXISTS ecommerce_catalog_prod;
```

## MAKING DYNAMIC USING WIDGETS 

```python
dbutils.widgets.text("catalog","ecommerce_catalog_dev")
catalog = dbutils.widgets.get("catalog")
```

## DYNAMIC SCHEMA CREATION

```python
%sql
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).bronze;
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).silver;
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).gold;
```

## DYNAMIC VOLUME CREATION

```python
%sql
CREATE VOLUME IF NOT EXISTS IDENTIFIER(:catalog).bronze.raw_data_volume
```
